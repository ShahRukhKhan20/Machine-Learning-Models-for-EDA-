```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.ensemble import AdaBoostRegressor
```

```python
df = pd.read_excel("FinTech_ChainAudit.xlsx")
```

```python
df["Sales"] = pd.to_numeric(df["Sales"], errors="coerce")
```

```python
df[df.select_dtypes(include="number").columns] = (
    df.select_dtypes(include="number")
      .fillna(df.select_dtypes(include="number").median())
)
```

```python
df[df.select_dtypes(include="object").columns] = (
    df.select_dtypes(include="object")
      .fillna("Unknown")
)
```

```python
df = df.drop_duplicates()
```

```python
df["Sales"].hist()

plt.show()
```

```python
sns.boxplot(x=df["Daily_Revenue_USD"])

plt.show()
```

```python
sns.scatterplot(
    x="Daily_Revenue_USD",
    y="Sales",
    data=df
)

plt.show()
```

```python
correlation = df.select_dtypes(include="number").corr()

sns.heatmap(correlation, annot=True)

plt.show()
```

```python
X = df[["Amount ", "Gas_Fee_USD", "Daily_Revenue_USD"]]

y = df["Sales"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
model = AdaBoostRegressor(
    n_estimators=100,
    learning_rate=1.0,
    random_state=42
)
```

```python
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
```

```python
mean_absolute_error(y_test, predictions)
```

```python
mean_squared_error(y_test, predictions)
```

```python
np.sqrt(mean_squared_error(y_test, predictions))
```

```python
r2_score(y_test, predictions)
```

```python
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

importance
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    importance["Feature"],
    importance["Importance"],
    color="skyblue"
)

plt.title("Feature Importance - AdaBoost Regressor")
plt.xlabel("Features")
plt.ylabel("Importance")

plt.show()
```

```python
plt.figure(figsize=(7,6))

plt.scatter(y_test, predictions, color="blue")

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    color="red",
    linewidth=2
)

plt.title("Actual vs Predicted Sales")
plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")

plt.show()
```

## INTERPRETATION
The AdaBoost Regressor model was used to predict Sales using the independent variables (Amount, Gas_Fee_USD, and Daily_Revenue_USD). The model's performance was evaluated using Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R² Score. Lower values of MAE, MSE, and RMSE indicate better prediction accuracy, while a higher R² Score indicates better model performance. AdaBoost Regressor combines multiple weak decision tree models, where each new model focuses on improving the errors made by the previous ones. The Feature Importance analysis identifies the variables that have the greatest influence on predicting Sales. AdaBoost is particularly effective for improving prediction accuracy compared to a single decision tree by sequentially learning from previous prediction errors.
