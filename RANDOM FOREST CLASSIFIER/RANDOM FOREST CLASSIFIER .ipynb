```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
df = pd.read_excel("New_columns.xlsx")
```

```python
df["Sales"] = pd.to_numeric(df["Sales"], errors="coerce")
```

```python
df[df.select_dtypes(include="number").columns] = (
    df.select_dtypes(include="number")
      .fillna(df.select_dtypes(include="number").median())
)
```

```python
df[df.select_dtypes(include="object").columns] = (
    df.select_dtypes(include="object")
      .fillna("Unknown")
)
```

```python
df = df.drop_duplicates()
```

```python
df["Sales"].hist()
plt.show()
```

```python
sns.boxplot(x=df["Daily_Revenue_USD"])
plt.show()
```

```python
sns.scatterplot(
    x="Daily_Revenue_USD",
    y="Sales",
    data=df
)
plt.show()
```

```python
correlation = df.select_dtypes(include="number").corr()

sns.heatmap(correlation, annot=True)

plt.show()
```

```python
df["Sales_Category"].value_counts().plot.pie(
    autopct="%1.1f%%"
)

plt.show()
```

```python
X = df[["Amount", "Gas_Fee_USD", "Daily_Revenue_USD"]]

y = df["Sales_Category"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
model = RandomForestClassifier(random_state=42)
```

```python
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
```

```python
accuracy_score(y_test, predictions)
```

```python
cm = confusion_matrix(y_test, predictions)

pd.DataFrame(
    cm,
    index=["Actual 0", "Actual 1"],
    columns=["Predicted 0", "Predicted 1"]
)
```

```python
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
report = classification_report(
    y_test,
    predictions,
    output_dict=True
)

pd.DataFrame(report).transpose()
```

```python
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance.sort_values(
    by="Importance",
    ascending=False
)
```

## INTERPRETATION 

The Random Forest Classifier model was used to classify transactions into Low Sales (0) and High Sales (1) based on the predictor variables (Amount, Gas_Fee_USD, and Daily_Revenue_USD). The model's performance was evaluated using Accuracy, Confusion Matrix, and Classification Report. A higher accuracy indicates better classification performance, while the confusion matrix shows the number of correct and incorrect predictions for each class. The classification report provides Precision, Recall, F1-Score, and Support to evaluate the model in detail. Unlike Logistic Regression, Random Forest Classifier also provides Feature Importance, which identifies the variables that contribute the most to the classification process. Because it combines predictions from multiple decision trees, Random Forest Classifier generally produces more robust and accurate results than a single Decision Tree Classifier, especially on complex datasets.
